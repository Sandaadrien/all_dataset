```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression as SklearnLR
```

## Preparing the Data

```python
column_features=[
    #"online_booking",
    "photo_count",
    "has_plan",
    "has_virtual_tour",
    "has_video",
    "has_home_staging",
    "bedrooms",
    "surface_m2"
    #"property_tags"
]
num_cols=["photo_count","bedrooms","surface_m2"]
bool_cols=["has_plan","has_virtual_tour","has_video","has_home_staging"]
```

```python
df=pd.read_csv('data.csv',usecols=column_features+["price"])
```

```python
y=np.log(df['price'])
X=df[column_features].copy()
X[num_cols]=X[num_cols].fillna(0)
X[bool_cols]=X[bool_cols].fillna(False).astype(int)
```

```python
np.random.seed(42)
indices = np.arange(len(X))
np.random.shuffle(indices)


train_size = int(0.8 * len(X))
train_idx = indices[:train_size]
test_idx = indices[train_size:]

X_train = X.iloc[train_idx]
y_train = y.iloc[train_idx]
X_test  = X.iloc[test_idx]
y_test  = y.iloc[test_idx]
```

## The LinearRegression Class hand made

```python
class LinearRegression:
    def __init__(self):
        self.w=None

    def train(self, X, y):        
        X_b=np.c_[np.ones((X.shape[0],1)),X]
        self.w=np.linalg.pinv(X_b.T.dot(X_b)).dot(X_b.T.dot(y))

    def predict(self, X):
        X_b=np.c_[np.ones((X.shape[0],1)),X]
        return np.dot(X_b,self.w)
```

## Use of the LinearRegression hand made:

```python
lr=LinearRegression()
lr.train(X_train,y_train)
y_predict=lr.predict(X_test)
```

## Use of the SkLearn

```python
sklearn=SklearnLR()
sklearn.fit(X_train,y_train)
y_sklearn=sklearn.predict(X_test)
```

## Showing the result

```python
def mse(y_true,y_predict):
    
    return np.mean((y_true-y_predict)**2)

comparison=pd.DataFrame({
    "y_true":y_test,
    "y_predict":y_predict,
    "y_sklearn":y_sklearn
})
print(f"\nPrediction Loss: {mse(y_test,y_predict)}\n")
print("Comparaison des valeurs:")
comparison.head(20)
```
